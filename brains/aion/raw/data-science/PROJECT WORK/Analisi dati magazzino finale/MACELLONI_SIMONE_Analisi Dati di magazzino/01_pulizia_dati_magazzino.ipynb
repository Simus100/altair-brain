# Pulizia e preparazione del dataset di magazzino

In questo notebook viene eseguita la pulizia iniziale del dataset di magazzino.

L'obiettivo è correggere eventuali problemi nei dati, controllare valori mancanti e duplicati, uniformare i testi, convertire correttamente prezzi e date e preparare un dataset pulito da utilizzare nelle analisi successive.

```python
#Importazione delle librerie per l'elaborazione dei dati:

import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Verifica della cartella di lavoro e presenza del dataset originale:

file_input = Path("Dataset.csv")

print("Cartella di lavoro:")
print(Path.cwd())

print("\nPercorso del dataset:")
print(file_input.resolve())

print("\nDataset presente?")
print(file_input.exists())
```

```python
# Importazione del dataset:

df = pd.read_csv(file_input)

print("Dataset importato.")
print("\nPrime 10 righe del dataset:")

df.head(10)
```

```python
# Controllo dimensioni del dataset

print("Numero di righe e colonne:")
print(df.shape)

print("\nNomi delle colonne:")
print(df.columns.tolist())
```

```python
# Controllo della struttura e dei tipi di dato

df.info()
```

---

E' stato individuato un errore di nomenclatura dell attributo categoria, erronemanete definito: "Catagory"



```python
# Correzione del nome della colonna "Category"

df = df.rename(columns={"Catagory": "Category"})

print("Nomi delle colonne:")
print(df.columns.tolist())
```

```python
# Controllo dei valori mancanti per ogni colonna

valori_mancanti = df.isna().sum()

print("Valori mancanti per colonna:")
print(valori_mancanti)
```


---

E' stato identificato un potenziale valore mancante nella colonna delle categorie. Di seguito verrà analizzato nel dettaglio questo elemento


```python
# Visualizzazione delle righe che contengono almeno un valore mancante

righe_con_valori_mancanti = df[df.isna().any(axis=1)]

print("Righe con almeno un valore mancante:")
righe_con_valori_mancanti
```

```python
# Elenco delle categorie presenti nel dataset

print("Categorie presenti:")
print(df["Category"].value_counts(dropna=False))
```

```python
# Assegnazione della categoria corretta al prodotto Cabbage

# df.loc[righe, colonne] -> Forma canonica di selezione nel df
 
df.loc[df["Product_Name"] == "Cabbage", "Category"] = "Fruits & Vegetables"
```

```python
# Controllo della correzione

df.loc[
    df["Product_Name"] == "Cabbage",
    ["Product_ID", "Product_Name", "Category"]
]
```

```python
# Controllo delle righe duplicate

numero_duplicati = df.duplicated().sum()

print("Numero di righe duplicate:")
print(numero_duplicati)
```

```python
# Controllo dei Product_ID duplicati

duplicati_product_id = df["Product_ID"].duplicated().sum()

print("Numero di Product_ID duplicati:")
print(duplicati_product_id)
```

```python

print("Valori mancanti in Category:")
print(df["Category"].isna().sum())
```

```python
# Elenco  dei valori unici

print("Categorie:")
print(sorted(df["Category"].dropna().unique()))

print("\nStati:")
print(sorted(df["Status"].dropna().unique()))


# Controllo dei valori presenti nelle colonne categoriche

print("Valori presenti nella colonna Category:")
print(df["Category"].value_counts(dropna=False))

print("\nValori presenti nella colonna Status:")
print(df["Status"].value_counts(dropna=False))
```

```python
# Pulizia di eventuali spazi iniziali o finali nelle colonne categoriche

colonne_testo = [
    "Product_ID",
    "Product_Name",
    "Category",
    "Supplier_ID",
    "Supplier_Name",
    "Warehouse_Location",
    "Status"
]

for colonna in colonne_testo:
    df[colonna] = df[colonna].str.strip()

# Verifica successiva alla pulizia

print("Valori presenti in Category:")
print(df["Category"].value_counts(dropna=False))

print("\nValori presenti in Status:")
print(df["Status"].value_counts(dropna=False))
```

```python
# Controllo iniziale della colonna Unit_Price

print("Tipo di dato di Unit_Price:")
print(df["Unit_Price"].dtype)

print("\nPrimi 5 valori:")
print(df["Unit_Price"].head(5))

print("\nValori mancanti:")
print(df["Unit_Price"].isna().sum())
```

```python
# Rimozione di $ e conversione della colonna Unit_Price in formato numerico

df["Unit_Price"] = pd.to_numeric(df["Unit_Price"].astype(str).str.replace("$", "", regex=False).str.strip(), errors="coerce")

# Verifica del tipo di dato e di eventuali valori mancanti

print("Tipo di dato:", df["Unit_Price"].dtype)
print("Conversioni non riuscite:", df["Unit_Price"].isna().sum())
print(df["Unit_Price"].head(10))
```

```python
# Controllo iniziale delle colonne contenenti date

print("Tipo di dato di Date_Received:")
print(df["Date_Received"].dtype)

print("\nTipo di dato di Last_Order_Date:")
print(df["Last_Order_Date"].dtype)

print("\nTipo di dato di Expiration_Date:")
print(df["Expiration_Date"].dtype)

print("\nPrime 20 righe delle colonne data:")
print(df[["Date_Received", "Last_Order_Date", "Expiration_Date"]].head(20))
```

```python
# Conversione delle colonne data dal formato testo MM/DD/YYYY al formato datetime

colonne_data = ["Date_Received", "Last_Order_Date", "Expiration_Date"]

for colonna in colonne_data:
    df[colonna] = pd.to_datetime(df[colonna], format="%m/%d/%Y", errors="coerce")
    
```

```python
# Verifica del tipo di dato e di eventuali errori di conversione nelle colonne data

print(df[colonne_data].dtypes)

print("\nValori mancanti dopo la conversione:")
print(df[colonne_data].isna().sum())

print("\nPrime 20 righe:")
print(df[colonne_data].head(20))
```

```python
# Controllo delle possibili incoerenze tra ricezione e scadenza

# Righe in cui la scadenza è precedente alla ricezione

date_non_coerenti = df[
    df["Expiration_Date"] < df["Date_Received"]
]

numero_non_coerenti = len(date_non_coerenti)

#risultato

print("Totale righe:", len(df))
print("Possibili incoerenze tra ricezione e scadenza:", numero_non_coerenti)
```

---
Sono state individuate 496 potenziali incoerenze temporali, nelle quali la data di scadenza precede la data dell’ultimo ricevimento. Poiché il dataset non contiene un identificativo del lotto, questi record vengono segnalati per verifica e non corretti.

```python
# Controllo dei valori numerici non validi o poco plausibili

# Selezioniamo le righe con quantità di magazzino negative

stock_negativo = df[df["Stock_Quantity"] < 0]

# Selezioniamo le righe con livello di riordino negativo

riordino_negativo = df[df["Reorder_Level"] < 0]

# Selezioniamo le righe con quantità da riordinare negativa

quantita_riordino_negativa = df[df["Reorder_Quantity"] < 0]

# Selezioniamo le righe con prezzo nullo o negativo

prezzo_non_valido = df[df["Unit_Price"] <= 0]

# Risultati

print("Stock negativo:", len(stock_negativo))
print("Livello di riordino negativo:", len(riordino_negativo))
print("Quantità di riordino negativa:", len(quantita_riordino_negativa))
print("Prezzo nullo o negativo:", len(prezzo_non_valido))
```

```python
# Verifica finale del dataset prima dell'esportazione

print("DIMENSIONI DEL DATASET")
print("Righe:", df.shape[0])
print("Colonne:", df.shape[1])

print("\nTIPI DI DATO")
print(df.dtypes)

print("\nVALORI MANCANTI")
print(df.isna().sum())

print("\nRIGHE DUPLICATE")
print(df.duplicated().sum())

print("\nDUPLICATI NEL PRODUCT_ID")
print(df["Product_ID"].duplicated().sum())
```

## Prime osservazioni sulle variabili numeriche

Le principali variabili quantitative mostrano distribuzioni abbastanza regolari, con valori medi e mediani generalmente vicini.

Le quantità di stock, i livelli di riordino, le quantità di riordino, il volume delle vendite e il tasso di rotazione presentano intervalli ampi. Dalla sola tabella descrittiva non emergono valori impossibili, ma le distribuzioni verranno approfondite con analisi specifiche.

La variabile `Unit_Price` mostra una maggiore asimmetria: il prezzo medio è superiore alla mediana e il valore massimo è molto distante dal terzo quartile. Questo suggerisce la presenza di alcuni prodotti con prezzo unitario molto più elevato rispetto alla maggior parte del catalogo.

Le colonne temporali coprono un intervallo di circa un anno, da febbraio 2024 a febbraio 2025.

```python
# Calcolo dei limiti per individuare i possibili outlier di Unit_Price

Q1 = df["Unit_Price"].quantile(0.25)
Q3 = df["Unit_Price"].quantile(0.75)

IQR = Q3 - Q1

limite_inferiore = Q1 - 1.5 * IQR
limite_superiore = Q3 + 1.5 * IQR

print("Primo quartile:", round(Q1, 2))
print("Terzo quartile:", round(Q3, 2))
print("Intervallo interquartile:", round(IQR, 2))
print("Limite inferiore:", round(limite_inferiore, 2))
print("Limite superiore:", round(limite_superiore, 2))
```

```python
# Selezione dei prodotti con prezzo superiore alla soglia statistica

outlier_prezzo = df[
    df["Unit_Price"] > limite_superiore
].copy()

print("Numero di prodotti con prezzo estremo:", len(outlier_prezzo))
```

```python
# Visualizzazione dei prodotti con prezzo superiore alla soglia statistica

display(
    outlier_prezzo[
        [
            "Product_ID",
            "Product_Name",
            "Category",
            "Unit_Price"
        ]
    ]
    .sort_values(
        by="Unit_Price",
        ascending=False
    )
    .head(20)
)
```

### Interpretazione

Il metodo IQR individua come possibili valori estremi i prezzi superiori al limite calcolato.

Questi valori non vengono eliminati, perché possono rappresentare prodotti realmente più costosi. Vengono mantenuti e segnalati per le analisi successive.
